```python
# This small snypt of code is for calculating what is the max thickness of a thin foil to be invisible to UT
# Assumption: the foil must be ~ 1/1000 of the wavelength in the metal to be invisible.
# This was created to support the microtexturing project when trying to determine foil thickness to possibly use as fiducial markers
```

```python
# import the necessary libraries
import math
```

```python
# set the independent variables
frac = 1/1000 #set the fraction of the wavelength the foil must be
Foil = "Gold" #identify the material being used as the foil
c = 3.24E+5   # [cm/s] set the longitudinal speed of sound in the metal. Either in bar or bulk should work
f = 10E+6     # [Hz] set the frequency of the transducer
```

```python
# calculate the wavelenth and max foil thickness

wave_cm = c/f
wave_um = wave_cm * 10_000
thickness_um = frac*wave_um
thickness_nm = 1000*thickness_um

print(thickness_nm)

```
