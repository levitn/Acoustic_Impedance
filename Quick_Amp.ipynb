```python
# This code is for doing quick calculation of acoustic impedance of a material
# It requires knowing the modulus and density to calculate the velcity c_0
# It requires knowing the poison's ration and using it in conjuction with c_0 to calculate the bulk veloctiy c_B
# It the require multiplying the density by c_B to obtain the characteristic impedance
# equations Referenced from Ultrasonics: Fundamentals, Technologies, and Applications
```

```python
# import necessary libraries
import math
```

```python
# set variables for Material 1

Material_1 = "Titanium" #identify the material being calculated 
Elastic_1 = 110 # [GPa] identify the elastic modulus
roh_1 = 4_510 # [kg/m^3] identify the density
poi_1 = 0.33 #identify the Poison's ratio

```

```python
# calculate the longitudinal wave speeds and the acoustic impedance for Material 1
c_0_1 = math.sqrt(Elastic_1*1E+9/roh_1) 
c_B_1 = c_0_1*math.sqrt((1-poi_1)/((1+poi_1)*(1-2*poi_1)))
Zimp_1 = c_B_1*roh_1/1E+6 #this outputs the impedance in units of MRayl 1E+6*[kg/m^2-s]
print(f"The acoustic impedance of {Material_1} is {round(Zimp_1, 4)} MRayl")
```
