```python
# This code is for doing quick calculation of acoustic impedance of a material
# It requires knowing the modulus and density to calculate the velcity c_0
# It requires knowing the poison's ration and using it in conjuction with c_0 to calculate the bulk veloctiy c_B
# It the require multiplying the density by c_B to obtain the characteristic impedance
```

```python
# import necessary libraries
import math
```

```python
# set variables

Material = "Hafnium" #identify the material being calculated 
E = 130 # [GPa] identify the elastic modulus
roh = 13_310 # [kg/m^3] identify the density
poi = 0.33 #identify the Poison's ratio

```

```python
# calculate the longitudinal wave speeds and the acoustic impedance
c_0 = math.sqrt(E*1E+9/roh) 
c_B = c_0*math.sqrt((1-poi)/((1+poi)*(1-2*poi)))
Z = c_B*roh/1E+6 #this outputs the impedance in units of MRayl 1E+6*[kg/m^2-s]
print(f"The acoustic impedance of {Material} is {round(Z, 4)} MRayl")
```
